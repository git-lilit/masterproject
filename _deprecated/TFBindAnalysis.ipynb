```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the data
data_folder = "../../../data"  # Replace with the actual path
x_data = np.load(f"{data_folder}/tf_bind_8-x-0.npy")
y_data = np.load(f"{data_folder}/tf_bind_8-y-0.npy")

# Visualize the distribution of y_data
plt.figure(figsize=(10, 6))

# Histogram
plt.subplot(1, 2, 1)
plt.hist(y_data, bins=20, color='skyblue', edgecolor='black', alpha=0.7)
plt.title("Histogram of y_data")
plt.xlabel("Values")
plt.ylabel("Frequency")

# Kernel Density Estimation (KDE) Plot
plt.subplot(1, 2, 2)
sns.kdeplot(y_data, color='darkblue', fill=True, alpha=0.6)
plt.title("KDE Plot of y_data")
plt.xlabel("Values")
plt.ylabel("Density")

plt.tight_layout()
plt.show()
```

```python
import numpy as np

# Example data and threshold

# Proportions
proportion_below = np.sum(y_data < 0.225) / len(y_data)
proportion_above = np.sum(y_data > 0.8) / len(y_data)

print(f"Proportion of values below {0.25}: {proportion_below:.2f}")
print(f"Proportion of values above {0.8}: {proportion_above:.2f}")
```

```python
import numpy as np


# Calculate the 10th and 90th percentiles
lower_bound = np.percentile(y_data, 10)
upper_bound = np.percentile(y_data, 90)

# Exclude the bottom 10% and the upper 10%
filtered_data = y_data[(y_data >= lower_bound) & (y_data <= upper_bound)]

print(f"Lower bound (10th percentile): {lower_bound:.2f}")
print(f"Upper bound (90th percentile): {upper_bound:.2f}")
print(f"Number of remaining values: {len(filtered_data)/len(y_data)}")
```

```python
import numpy as np

# Example data
np.random.seed(42)  # For reproducibility

# Calculate the 20th and 80th percentiles
lower_bound = np.percentile(y_data, 15)
upper_bound = np.percentile(y_data, 85)

# Filter for the middle 60%
middle_mask = (y_data >= lower_bound) & (y_data <= upper_bound)
y_filtered = y_data[middle_mask]
x_filtered = x_data[middle_mask.squeeze(1), :]

sample_size = int(0.7 * len(y_filtered))
indices = np.random.choice(len(y_filtered), size=sample_size, replace=False)

x_sampled = x_filtered[indices]
y_sampled = y_filtered[indices]

# Print results to verify
print(f"Original x_data shape: {x_data.shape}")
print(f"Original y_data shape: {y_data.shape}")
print(f"Filtered x_data shape: {x_filtered.shape}")
print(f"Filtered y_data shape: {y_filtered.shape}")
print(f"Sampled x_data shape: {x_sampled.shape}")
print(f"Sampled y_data shape: {y_sampled.shape}")
```
