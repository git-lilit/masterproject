```python
import torch
from holo.test_functions.closed_form import Ehrlich
```

```python
test_fn_params = {
    "num_states": 4,
    "dim": 64,
    "num_motifs": 4,
    "motif_length": 4,
    "quantization": 4,
    "random_seed": 0,
}

device = torch.device("cuda")

test_fn = Ehrlich(**test_fn_params, negate=False)

seq_len = test_fn_params["dim"]
num_tokens = test_fn_params["num_states"]
start_token = num_tokens

model_params = {
    "embedding_dim": 32,
    "num_heads": 4,
    "output_dim": test_fn_params["num_states"],
    "vocab_size": test_fn_params["num_states"] + 2,
}

test_fn.to(device, torch.float)
```

```python
from lib.training import sample_sequences
from lib.NewSequenceModel import SequenceModel
```

```python
q = SequenceModel(**model_params, seq_len=test_fn_params["dim"])
```

```python
fixed_state_batch = sample_sequences(
    q=q,  # Use a random policy (or disable Q-function dependence)
    epsilon=1.0,  # Ensure full exploration if required
    batch_size=64,
    start_token=start_token,
    seq_len=seq_len,
    num_states=num_tokens,
    device=device
)
```

```python
fixed_state_batch.shape

fixed_state_batch
```

```python
test_fn.to(device, torch.float)
test_fn(fixed_state_batch)
```

```python
sum(test_fn(fixed_state_batch))
```
